## Importing Library

```python
# We are using just numpy for this numerical analysis
import numpy as np
```

## Data Preparation

```python
# Data from the table
temperature = np.array([40, 40, 40, 30, 15, 15, 15, 30, 15, 30, 30, 30, 40, 30])
humidity = np.array([90, 90, 90, 90, 70, 70, 70, 90, 70, 70, 70, 90, 70, 90])
```

```python
# Combine into a 2D array (14 samples, 2 features)
data = np.column_stack((temperature, humidity))
```

## Calculating for Mean

```python
mean_vec = np.mean(data, axis=0)
print(f"   Mean Temperature : {mean_vec[0]:.2f}")
print(f"   Mean Humidity    : {mean_vec[1]:.2f}\n")
```

## Calculting Mean Deviation (Anomalies)

```python
mean_dev = data - mean_vec   
print("   Mean deviations:")
print(np.round(mean_dev, 2))
print()
```

## Finding the Variance

```python
var = np.var(data, axis=0, ddof=0)
print(f"   Var(Temperature) : {var[0]:.2f}")
print(f"   Var(Humidity)    : {var[1]:.2f}\n")
```

## Calculating the Standard Deviation

```python
std = np.std(data, axis=0, ddof=0)
print(f"   Std(Temperature) : {std[0]:.2f}")
print(f"   Std(Humidity)    : {std[1]:.2f}\n")
```

## Finding the Covariance

```python
cov_temp_hum = np.cov(temperature, humidity, ddof=0)[0,1]

print(f"   Covariance(Temp, Hum): {cov_temp_hum:.2f}\n")
```

## The Covariance Matrix

```python
cov_matrix = np.cov(data.T, ddof=0)

print(np.round(cov_matrix, decimals=2))
print()
```

## Finding the Eigenvalues

```python
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

print("   Eigenvalues:", np.round(eigenvalues, decimals=4))

```

## Finding the Eigenvectors

```python
print("   Eigenvectors:")

print(np.round(eigenvectors, decimals=4))
print()
```

## Normalizing the Eigenvectors

```python
norms = np.linalg.norm(eigenvectors, axis=0)           # Get the magnitude of each eigenvector
normalized_eigenvectors = eigenvectors / norms         # Divide each vector by its magnitude
```

```python
print("   Normalized Eigenvectors:")
print(np.round(normalized_eigenvectors, decimals=4))
```

## Principal Components

```python
principal_components = mean_dev @ normalized_eigenvectors

print("\n   (PC1, PC2):")
print(np.round(principal_components, decimals=2))
```

## Explanation


#### Key Observations:

- The data points naturally form **distinct clusters** in the new principal component space.
- Points with **identical original values** (multiple `(40, 90)` entries) have the **same PC scores** (`-1.62, -15.1`), which is expected and correct.
- There is a clear separation between:
  - High temperature + high humidity points
  - Low temperature + low humidity points

#### Interpretation:

- **PC1** (first column) primarily captures the **temperature variation**. Higher PC1 scores generally correspond to higher temperatures.
- **PC2** (second column) captures the remaining variation, mostly related to **humidity differences**.
- The strong clustering shows that **temperature and humidity are highly correlated** in this dataset; as temperature increases, humidity tends to increase as well.

#### Conclusion

This PCA analysis confirms that **temperature has a strong influence on humidity**. 

By transforming the data into principal components, we saw the main patterns more clearly. The first principal component (PC1) explains the majority of the variance, meaning most of the information in the dataset can be represented along a single dominant direction.

This supports the idea that **temperature is a major driving factor** of humidity levels in the given data.
