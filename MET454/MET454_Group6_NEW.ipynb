**Question:**

Consider the differential equation 
$$
\dfrac{dy}{dx} = x + 2y
$$

Suppose $y = s(x)$ satisfies the initial condition $s(1) = m$, where $m$ is a constant. 

Using Euler’s method with step size $h = 0.5$, starting from $x = 1$, the approximation obtained is $s(2) = 15.5$. 

**Determine the value of $m$.**

### What is Euler’s Method?

Euler’s method is a simple numerical technique to approximate the solution of a differential equation. 

It is based on the idea that the derivative $\frac{dy}{dx}$ gives the **slope of the tangent line** at any point. We move forward in small steps of size $h$ along this tangent line to estimate the next value of $y$.

**Formula:**
$$
y_{n+1} = y_n + h \cdot f(x_n, y_n)
$$

where $f(x, y) = \frac{dy}{dx}$.

**Step 1:** From $x=1$ to $x=1.5$

Given:
- $x_0 = 1$, $y_0 = m$

Slope at this point:
$$
f(x_0, y_0) = x_0 + 2y_0 = 1 + 2m
$$

Next $y$-value:
$$
y_1 = y_0 + h \cdot f(x_0, y_0) = m + 0.5(1 + 2m)
$$

Simplifying:
$$
y_1 = m + 0.5 + m = 2m + 0.5
$$

Now we are at: $x_1 = 1.5$, $y_1 = 2m + 0.5$

## Step 1: In Python

```python
# Import the SymPy library, which is used for symbolic mathematics.
# It allows us to work with the unknown initial value 'm' symbolically instead of assigning it a fixed numerical value.

import sympy as sp

# Define the function f(x, y)
print("Differential Equation: dy/dx = x + 2y") # This represents the rate of change of y with respect to x.
print("Step size h = 0.5")                     # A step size of 0.5 means that we move along the x-axis in intervals of 0.5.
print("Starting point: x0 = 1, y0 = m")        # x0 represents the initial x-value.
                                               # y0 is represented by 'm', meaning the starting y-value is unknown.
print("Target: Approximate y(2) = 15.5\n")     # We want to find the initial value 'm' that will make the approximation reach y(2) = 15.5

# Define symbolic variable for m
m = sp.symbols('m')

# Euler's Method (First Step (x=1 to x=1.5))
x0 = 1.0 
y0 = m
h = 0.5
f0 = x0 + 2 * y0

y1 = y0 + h * f0
x1 = x0 + h

print("First Step:")
print(f"   x0 = {x0}, y0 = m")
print(f"   f(x0,y0) = {f0}")                          # Display the slope calculated from the differential equation.
print(f"   y1 = {sp.simplify(y1)}  (at x1 = {x1})\n") # sp.simplify() is used to simplify the symbolic expression
```

**Step 2:** From $x=1.5$ to $x=2.0$

Slope at the new point:
$$
f(x_1, y_1) = x_1 + 2y_1 = 1.5 + 2(2m + 0.5)
$$

Simplifying:
$$
f(x_1, y_1) = 1.5 + 4m + 1 = 4m + 2.5
$$

Final approximation:
$$
y_2 = y_1 + h \cdot f(x_1, y_1) = (2m + 0.5) + 0.5(4m + 2.5)
$$

Simplify:
$$
y_2 = 2m + 0.5 + 2m + 1.25 = 4m + 1.75
$$

## Step 2: In Python

```python
# Step 2
print("STEP 2: From x = 1.5 to x = 2.0")
f1 = x1 + 2 * y1
y2 = y1 + h * f1

print(f"   Slope f(x1,y1) = {sp.simplify(f1)}")
print(f"   y2 = {sp.simplify(y2)}\n")
```

According to the problem, this approximation equals 15.5:

$$
4m + 1.75 = 15.5
$$

**Solve for $m$:**

$$
4m = 15.5 - 1.75 = 13.75
$$

$$
m = \frac{13.75}{4} = 3.4375
$$

**Exact fraction form:**

$$
m = \frac{55}{16}
$$


**Final Answer:**

$$
\boxed{m = 3.4375}
$$

## Step 3: Solving for m in Python and verifying with the Equation

```python
# The previous Euler calculations produced y2 as an expression
# containing the unknown starting value 'm'.
# By setting y2 equal to 15.5, we can solve for the value of m
# that makes the Euler approximation reach the required target.
equation = sp.Eq(y2, 15.5)
print("Equation to solve:")
print(equation)

# SymPy will then rearrange the equation and find the value(s) of m
# that satisfy the condition y2 = 15.5.
solution = sp.solve(equation, m)
print("\nSolution:")

# Loop through the solutions returned by SymPy and displays the symbolic solution
for sol in solution:
    print(f"m = {sol}  (or {float(sol):.4f} as decimal)")

# Numerical verification with the found m

# Convert the symbolic solution into a numerical floating-point value.
m_val = float(solution[0])
print("")
print(f"Using m = {m_val}")

# Now we repeat Euler's method using the calculated numerical value of m.
# This checks whether the obtained initial condition actually produces
# y(2) ≈ 15.5.

# Recompute with numerical value
y0_num = m_val                         # Replace the symbolic starting value with the numerical value found
f0_num = x0 + 2 * y0_num               # Calculate the first slope using the numerical starting value
y1_num = y0_num + h * f0_num           # Perform the first Euler step which gives us y(1.5)
f1_num = x1 + 2 * y1_num               # Calculate the slope at the second point
y2_num = y1_num + h * f1_num           # Perform the second Euler step which gives us y(2)

print(f"Step 1: y(1.5) ≈ {y1_num:.4f}")
print(f"Step 2: y(2)   ≈ {y2_num:.4f} ")
```

## Exact Solution + Error Calculatiion

```python
# Define x as a symbolic variable and y as a symbolic function of x
#Instead of treating y as a simple variable, we define it as a function
# The problem involves a differential equation where y changes with x.
x = sp.symbols('x')
y_func = sp.Function('y')(x)

# Define the differential equation
de = sp.Eq(y_func.diff(x), x + 2*y_func) # y_func.diff(x) represents the derivative of y with respect to x.
                                         # sp.Eq() creates an equation object that SymPy can solve.

# Solve the DE with initial condition y(1) = m_val
# SymPy will find the exact mathematical solution function y(x)
# that satisfies both the differential equation and the initial condition
exact_sol = sp.dsolve(de, y_func, ics={y_func.subs(x,1): m_val})

# Substitute x = 2 into the exact solution to find the actual value of y(2)
y2_exact = float(exact_sol.rhs.subs(x, 2).evalf())

print("EXACT SOLUTION:")
print("Exact solution y(x) =", exact_sol.rhs)             # Print the complete equation for y(x)
print(f"\nExact value at x=2 : y(2) ≈ {y2_exact:.4f}")    # Display the exact value of y at x = 2
print(f"Euler approximation: {15.5}")                     # This is the estimated value obtained using the numerical method
print(f"Absolute Error     : {abs(15.5 - y2_exact):.4f}") # Calculate the absolute error between the Euler approximation and the exact analytical solution
```

## Plotting the Euler vs Exact Solution

```python
# importing libraries for plotting
import numpy as np
import matplotlib.pyplot as plt

# Create the figure and set the sizee
plt.figure(figsize=(12, 8))

# Plot Exact Solution Curve
x_exact = np.linspace(0.7, 2.3, 400) # Generate many x-values between 0.7 and 2.3.
                                     # These points will be used to draw a smooth curve representing the exact analytical solution.


# Convert the symbolic exact solution obtained from SymPy into a
# numerical function that can accept NumPy arrays.
y_exact = sp.lambdify(x, exact_sol.rhs, 'numpy')(x_exact) # sp.lambdify() changes the symbolic expression y(x) into a normal Python function that can calculate many y-values quickly.
                                                          # The function is then evaluated for all x-values created above.




plt.plot(x_exact, y_exact, 'b-', linewidth=2.8, label='Exact Solution') # Plot the exact solution curve with a blue solid line.

# Plot Euler's Tangent Lines
plt.plot([1.0, 1.5], [m_val, m_val + 0.5*(1 + 2*m_val)], 
         'r-', linewidth=3, label='Euler Step 1')
plt.plot([1.5, 2.0], [m_val + 0.5*(1 + 2*m_val), 15.5], 
         'r-', linewidth=3, label='Euler Step 2')

# Scatter points show the actual locations calculated by Euler's method
# s controls the marker size.
# zorder places the points above the lines so they are clearly visible.
plt.scatter([1.0, 1.5, 2.0], [m_val, m_val + 0.5*(1 + 2*m_val), 15.5], 
            color='red', s=130, zorder=5, label='Euler Points')

# Clear Annotations

# A text box showing the starting point information is added with # plt.annotate()
# xy specifies the point being labeled.
# xytext controls where the text box appears relative to the point.
#  textcoords defines the coordinate system used for xytext.
# 'offset points' means:
    # The xytext values are treated as movement away from the xy point
    # rather than actual x and y coordinates on the graph.
# arrowprops draws an arrow pointing to the location.

plt.annotate('Start\nx=1.0\ny=m=3.4375', xy=(1.0, m_val), xytext=(-90, 30),
             textcoords='offset points', fontsize=11, fontweight='bold',
             bbox=dict(boxstyle="round,pad=0.5", facecolor="yellow", alpha=0.95),
             arrowprops=dict(arrowstyle='->'))

plt.annotate('After Step 1\nx=1.5\ny≈7.0625', xy=(1.5, m_val + 0.5*(1 + 2*m_val)), 
             xytext=(25, 35), textcoords='offset points', fontsize=11,
             bbox=dict(boxstyle="round,pad=0.5", facecolor="lightblue", alpha=0.95),
             arrowprops=dict(arrowstyle='->'))

plt.annotate('Final Approx\nx=2.0\ny=15.5', xy=(2.0, 15.5), xytext=(-125, -60),
             textcoords='offset points', fontsize=12, fontweight='bold',
             bbox=dict(boxstyle="round,pad=0.6", facecolor="lightgreen", alpha=0.95),
             arrowprops=dict(arrowstyle='->'))

plt.annotate(f'Exact y(2)≈{y2_exact:.2f}', xy=(2.0, y2_exact), xytext=(20, 20),
             textcoords='offset points', color='blue', fontsize=12, fontweight='bold',
             bbox=dict(boxstyle="round,pad=0.5", facecolor="white", alpha=0.9))

# Title and Labels
# The equation is written using LaTeX formatting for better mathematical display.
plt.title("Euler's Method vs Exact Solution\n" + 
          r"$\dfrac{dy}{dx} = x + 2y$, $\quad s(1)=m$, $\quad h=0.5$", 
          fontsize=16, pad=20)

plt.xlabel('x', fontsize=13)
plt.ylabel('y = s(x)', fontsize=13)
plt.grid(True, alpha=0.35)
plt.legend(fontsize=12)
plt.xlim(0.8, 2.2) # Set the visible x-axis range
plt.ylim(0, 32)    # Set the visible y-axis range

plt.show()
```

**Key Observation:**

There is a **significant difference** of approximately **14.19** between Euler’s approximation and the exact solution.

#### Why is the error so large?

- Euler’s method is a **first-order** numerical method, meaning its error is proportional to the step size $h$.
- In this problem, we used a relatively **large step size** ($h = 0.5$), covering the interval from $x=1$ to $x=2$ in only **two steps**.
- The true solution grows rapidly (it is convex upward), so using straight tangent lines over large intervals leads to **underestimation**.
- With such a coarse step size, the accumulated truncation error becomes substantial.

#### How to Improve Accuracy:

To get a better approximation, we must **reduce the step size** $h$. Smaller values of $h$ (e.g., $h=0.1$ or $h=0.01$) result in more steps and significantly lower error. 


---

**Conclusion:**

This example clearly demonstrates the **trade-off** in Euler’s method:  
**Larger step size: faster computation but lower accuracy**  
**Smaller step size: higher accuracy but more computational effort**
